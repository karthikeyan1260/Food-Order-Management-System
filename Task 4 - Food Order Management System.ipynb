```python
# Variables and Input
Customer_Name = input("Enter Customer Name:") #Name of the Customer in String
Age = int(input("Enter Age:")) #Age of the Customer in Integer
Food_Item = input("Enter Food Items:") #List the Food Items in List Method
Quantity = [] #List the Quantity in List Method
Quantity.append(int(input("Enter Quantity 1:"))) #Append Used to Add Value in List as Index
Quantity.append(int(input("Enter Quantity 2:")))
Quantity.append(int(input("Enter Quantity 3:")))
Quantity.append(int(input("Enter Quantity 4:")))

Price = []
Price.append(int(input("Enter Price 1:"))) #Append Used to Add Value in List as Index
Price.append(int(input("Enter Price 2:")))
Price.append(int(input("Enter Price 3:")))
Price.append(int(input("Enter Price 4:")))

Total = (Price[0]*Quantity[0] + Price[1]*Quantity[1] + Price[2]*Quantity[2] + Price[3]*Quantity[3]) #Multiplying the Indexes of Quantity and Price to get the Total
Home_Delivery = input("Home Delivery (True/False):")

# String handling
Customer_Name = Customer_Name.strip()

#List
Menu = ["Pizza", "Burger", "Briyani", "Chicken Lollipops", "Chicken Tandoori"] #Food Items ordered in Menu as List 

#Tuple
Delivery_Slots = ("12:00 PM - 1:00 PM", "2:00 PM - 3:00 PM", "7:00 PM - 8:00 PM") #Delivery Slots are Ordered in Tuple

#Sets
Payment_Methods = {"UPI", "Cash", "Card"} #Payment Methods are Arranged In Set
Selected_Payment = input("Select Payment Method:") #Customer May choose a Available Payment Methods

#Dictionary
Order_Details = {"Customer_Name" : Customer_Name, #Customer Name 
                 "Food_Item" : Food_Item, #Cutomer Ordered Food Items
                 "Quantity" : Quantity, #Quantity of the Food Item 
                 "Price" : Price, #Price of the Food Item
                 "Total" : Total, #Sum of the Price and Quantity
                 "Payment_Method" : Selected_Payment #Select a payment method For Pay
                }

#Dsiplay Customer Details
print("\n--- Customer Details ---")
print("Customer Name:",Customer_Name.upper()) #Upper is used to show the Customer Name in Capital Letters 
print("Age:",Age) 
print("Home Delivery:",Home_Delivery)

#Display Food Items
print("\n--- Food Items ---")
print("Food Item:",Food_Item.lower()) #Lower is used to show the Food Items in small letters
print("Food Name Lenghth:",len(Food_Item))

#Display Menu
print("\n--- Available Menus ---")
print(Menu)

#Delivery Slots
print("\n--- Delivery Slots ---")
print(Delivery_Slots)

#Display Payment Methods
print("\n--- Payment Methods ---")
print(Payment_Methods)
print("Payment Method Selected:",Selected_Payment)

#Display Order Summary
print("\n--- Order Summary ---")
print(Order_Details)

print("\n--- THANKS FOR THE ORDER. PLEASE VISIT AGAIN ---")
```
